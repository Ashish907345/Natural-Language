```python
with open('hi_part_1.txt', 'r', encoding='utf-8') as f:
    hindi_text = f.read()

print(len(hindi_text))

# Prepare the training data
training_corpus = [hindi_text[1:100000000]]
```

```python
from tokenizers import Tokenizer, models, trainers, pre_tokenizers, decoders, processors

# Initialize a BPE tokenizer
tokenizer = Tokenizer(models.BPE())

# Customize pre-tokenization and decoding
tokenizer.pre_tokenizer = pre_tokenizers.Whitespace()
tokenizer.decoder = decoders.BPEDecoder()


# Set up the trainer with a target vocabulary size
trainer = trainers.BpeTrainer(vocab_size=10000, special_tokens=["<pad>", "<s>", "</s>", "<unk>"])

# Train the tokenizer
tokenizer.train_from_iterator(training_corpus, trainer)

# Save the tokenizer
tokenizer.save("hindi_bpe_tokenizer.json")

# Function to calculate compression ratio
def compression_ratio(text, tokenized_output):
    original_length = len(text)
    tokenized_length = len(tokenized_output.tokens)

    print("Original Length:", original_length)
    print("Tokenized Length:", tokenized_length)
    return original_length / tokenized_length

# Example usage
example_text = "आपका स्वागत है"
encoded = tokenizer.encode(example_text)
print("Tokens:", encoded.tokens)
print("Compression Ratio:", compression_ratio(example_text, encoded))
```
